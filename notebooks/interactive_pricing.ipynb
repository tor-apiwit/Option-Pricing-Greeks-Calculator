```python
import sys
import os
# สั่งให้ Python รู้จักโฟลเดอร์แม่ (Root Directory) ก่อนทำการ Import
sys.path.append(os.path.abspath('..'))

%load_ext autoreload
%autoreload 2

import ipywidgets as widgets
from IPython.display import display, clear_output
from option_pricing_pkg.reports import price_report_with_vanilla, greek_report_with_vanilla
```

```python
def interactive_dashboard(S, K, T, r, sigma, option_type):
    # ล้างผลลัพธ์เก่าบนหน้าจอ
    clear_output(wait=True)
    
    print(f"=== Real-time Pricing Analysis: {option_type.upper()} Option ===")
    print(f"S={S}, K={K}, T={T} Yrs, r={r:.1%}, Sigma={sigma:.1%}\n")
    
    # ดึงฟังก์ชันเปรียบเทียบจากแพ็กเกจมาใช้งาน
    price_report_with_vanilla(S, K, T, r, sigma, option_type)
    print("\n")
    greek_report_with_vanilla(S, K, T, r, sigma, option_type)

# สร้าง Slider และเชื่อมโยงเข้ากับฟังก์ชัน
interactive_plot = widgets.interactive(
    interactive_dashboard,
    S=widgets.FloatSlider(value=100, min=50, max=150, step=1, description='Spot Price (S):'),
    K=widgets.FloatSlider(value=100, min=50, max=150, step=1, description='Strike (K):'),
    T=widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description='Time (Yrs):'),
    r=widgets.FloatSlider(value=0.015, min=0.0, max=0.10, step=0.001, readout_format='.1%', description='Risk-free (r):'),
    sigma=widgets.FloatSlider(value=0.32, min=0.05, max=1.0, step=0.01, readout_format='.1%', description='Volatility:'),
    option_type=widgets.Dropdown(options=['call', 'put'], value='call', description='Type:')
)

# แสดงผล Dashboard
display(interactive_plot)
```
